```python
import numpy as np
import pandas as pd
```

```python
movies_df = pd.read_excel('movies.xlsx')
ott_df = pd.read_csv('ott.csv')
print(movies_df.head())
print(movies_df.shape)
print(movies_df.dtypes)
print(ott_df.head())
print(ott_df.shape)
print(ott_df.dtypes)
```

```python
movies_num = movies_df.select_dtypes('number')
movies_num.head()
```

```python
movies_num = movies_num.drop('Year', axis=1)
movies_num.columns
```

```python
Q1 = movies_num.quantile(0.25)
Q3 = movies_num.quantile(0.75)
IQR = Q3 - Q1

IQR
```

```python
pd.DataFrame({'Q1': Q1, 'Q3': Q3, 'IQR': IQR})
```
