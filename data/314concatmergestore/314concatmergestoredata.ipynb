```python
import pandas as pd
tx_2010 = pd.read_csv('transactions_2010.csv')

print(tx_2010.shape)
tx_2010.head()
```

```python
# ways to determine the length of a datafram
print(len(tx_2010))

# how many rows, with a labelled output
print(f"{tx_2010.shape[0]} rows for tx_2010")
```

```python
tx_2011 = pd.read_csv('transactions_2011.csv')
print(tx_2011.shape)
tx_2011.head()
```

```python
tx_2010.dtypes
tx_2011.dtypes
tx_2010.info()
tx_2011.info()
tx_2010.head()
tx_2011.head()
```

```python
tx_2010['InvoiceDate'] = pd.to_datetime(tx_2010['InvoiceDate'], errors='coerce')
tx_2011['InvoiceDate'] = pd.to_datetime(tx_2011['InvoiceDate'], errors='coerce')
tx_2010.dtypes
tx_2011.dtypes
```

```python
# Combine the two DataFrames with the concat() function. 
transactions = pd.concat([tx_2010, tx_2011], axis=0) 

# Determine how many rows. 
print(f"{transactions.shape[0]} rows for all transactions") 
transactions.shape 
```

```python
# Read the CSV file from the current working directory. 
products = pd.read_csv('products.csv') 

# Use the left join to merge the two DataFrames. 
transactions_description = pd.merge(transactions, products, 
                                    on='StockCode', how='left') 

# View the new DataFrame. 
transactions_description.head()
```

```python
# Read the CSV file from the current working directory. 
customers = pd.read_csv('customers.csv') 

# Use the left join to merge the two DataFrames. 
transactions_description_country = pd.merge(transactions_description, 
                                            customers, 
                                            on='CustomerID', how='left') 

# View the new DataFrame. 
transactions_description_country.head()
```

```python
# Total sales by Country first qty by unit price

transactions_description_country['SaleTotal'] = (
    transactions_description_country['Quantity']
    * transactions_description_country['UnitPrice'])

# group by country and sum the sales

sales_by_country = (
    transactions_description_country
    .groupby('Country')
    .agg({'SaleTotal': ['sum', 'mean']})
    .sort_values(('SaleTotal', 'sum'), ascending=False)
)

sales_by_country
```

```python
# What is the top selling product?

# Group by product description, sum sales, sort, and read the top. 
top_products = (transactions_description_country 
                .groupby('Description')['SaleTotal'] 
                .agg('sum') 
                .sort_values(ascending=False)) 
top_products.head() 
```
