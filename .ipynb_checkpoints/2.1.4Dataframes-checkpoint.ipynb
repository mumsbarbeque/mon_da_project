```python
import pandas as pd
# Create an empty DataFrame and an empty Series.
df = pd.DataFrame()
ser = pd.Series(dtype='float64')

print(df)
print(ser)
```

```python
# Column title is the key, column data the value
data = {
    'Name': ['Tim', 'Sharon', 'Simone', 'Robert'],
    'Country': ['Australia', 'New Zealand', 'France', 'Thailand'],
    'Qualification': ['PhD', 'Diploma', 'MBA', 'BSc'],
    'Age': [47, 25, 30, 57]
}

# Build the DataFrame from the dictionary.
students = pd.DataFrame(data)

print(students)
```
