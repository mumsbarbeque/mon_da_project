```python
import pandas as pd

leave_records = {
    'remaining_sick' : [10, 8, 10, 9, 7, 10, 9, 10, 8, 5],
    'remaining_annual': [20, 20, 15, 16, 18, 15, 10, 12, 10, 20],
    'remaining_personal': [5, 4, 5, 0, 5, 2, 5, 4, 2, 5]}
employees = [215, 216, 217, 218, 219, 220, 221, 222, 223,224]
leave_cycle = pd.DataFrame(leave_records, index=employees)
leave_cycle.index.name = 'employee_ID'
print(leave_cycle)
```

```python
import pandas as pd

leave_data = {
    'Sick_leave_remaining': [10, 8, 10, 9, 7, 10, 9, 10, 8, 5],
    'Annual_leave_remaining': [20, 20, 15, 16, 18, 15, 10, 12, 10, 20],
    'Personal_leave_remaining': [5, 4, 5, 0, 5, 2, 5, 4, 2, 5]}
row_labels = [215, 216, 217, 218, 219, 220, 221, 222, 223, 224]
leave_cycle = pd.DataFrame(leave_data, index=row_labels)
leave_cycle.index.name = 'Personnel_ID'
print(leave_cycle)
```
