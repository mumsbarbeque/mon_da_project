```python
import pandas as pd
fitness = pd.read_csv('daily_activity.csv')
mylist = fitness.columns.values.tolist()
mylist
```

```python
# Add a column giving each row's distance from the mean.
fitness['cal_diff'] = fitness['Calories'] - fitness['Calories'].mean()
# Sense-check the new column.
print(fitness['cal_diff'])
print(fitness.head())
```

```python
print(fitness['Calories'].sort_values())
```
