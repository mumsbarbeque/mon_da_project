```python
import pandas as pd
fitness = pd.read_csv('daily_activity.csv')
print(fitness['Calories'].mean())
print(fitness['Calories'].median())
```

```python
# cumsum running total mechanical working only no presort

fitness['TotalSteps_cumulative'] = fitness['TotalSteps'].cumsum()
fitness[['TotalSteps', 'TotalSteps_cumulative']].head()
```

```python
# Correlation between calories and steps whole dataset

fitness[['TotalSteps', 'Calories']].corr()
```

```python
# min max scaling

fitness['VeryActiveMinutes_scaled'] = (
    fitness['VeryActiveMinutes'] - fitness['VeryActiveMinutes'].min()
)/(
    fitness['VeryActiveMinutes'].max() - fitness['VeryActiveMinutes'].min()
        )
fitness[['VeryActiveMinutes', 'VeryActiveMinutes_scaled']].head()
```

```python
# max with abs scalling

fitness['SedentaryMinutes_scaled'] = (
    fitness['SedentaryMinutes'] / fitness['SedentaryMinutes'].abs().max())
fitness[['SedentaryMinutes', 'SedentaryMinutes_scaled']].head()
```
