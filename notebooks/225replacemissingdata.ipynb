```python
import pandas as pd

fitness_raw = pd.read_csv('daily_activity_raw.csv')

fitness_raw.head()
```

```python
fitness_raw['TotalSteps'].isnull().sum()
```

```python
fitness_raw['TotalSteps'] = fitness_raw['TotalSteps'].fillna(fitness_raw['TotalSteps'].mean())
fitness_raw['TotalSteps'].isnull().sum()
```

```python
fitness_raw['TrackerDistance'].isnull().sum()
```

```python
fitness_raw['TrackerDistance'] = fitness_raw['TrackerDistance'].fillna(0)
fitness_raw['TrackerDistance'].isnull().sum()
```

```python
fitness_raw['LoggedActivitiesDistance'].isnull().sum()
```

```python
fitness_raw['LoggedActivitiesDistance'] = fitness_raw['LoggedActivitiesDistance'].fillna(0)
fitness_raw['LoggedActivitiesDistance'].isnull().sum()
```

```python
fitness_raw['VeryActiveDistance'].isnull().sum()
```

```python
fitness_raw['VeryActiveDistance'] = fitness_raw['VeryActiveDistance'].ffill()
fitness_raw['VeryActiveDistance'].isnull().sum()
```

```python
fitness_raw['LightActiveDistance'].isnull().sum()
```

```python
fitness_raw['LightActiveDistance'] = fitness_raw['LightActiveDistance'].bfill()
fitness_raw['LightActiveDistance'].isnull().sum()
```

```python
fitness_raw['VeryActiveMinutes'].isnull().sum()
```

```python
fitness_raw['VeryActiveMinutes'] = fitness_raw['VeryActiveMinutes'].min()
fitness_raw['VeryActiveMinutes'].isnull().sum()
```

```python
fitness_raw['FairlyActiveMinutes'].isnull().sum()
```

```python
fitness_raw['FairlyActiveMinutes'] = fitness_raw['FairlyActiveMinutes'].min()
fitness_raw['FairlyActiveMinutes'].isnull().sum()
```

```python
fitness_raw['LightlyActiveMinutes'].isnull().sum()
```

```python
fitness_raw['LightlyActiveMinutes'] = fitness_raw['LightlyActiveMinutes'].fillna(fitness_raw['LightlyActiveMinutes'].max())
fitness_raw['LightlyActiveMinutes'].isnull().sum()
```

```python
fitness_raw['SedentaryMinutes'].isnull().sum()
```

```python
fitness_raw['SedentaryMinutes'] = fitness_raw['SedentaryMinutes'].fillna(fitness_raw['SedentaryMinutes']).max()
fitness_raw['SedentaryMinutes'].isnull().sum()
```

```python
fitness_raw = fitness_raw.fillna(0)
fitness_raw[fitness_raw.isna().any(axis=1)]
```
