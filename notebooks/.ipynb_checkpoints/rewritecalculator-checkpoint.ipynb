```python
"""A simple calculator."""


def add(x, y):
    """Return the sum of x and y."""
    return x + y


def subtract(x, y):
    """Return x minus y."""
    return x - y


def multiply(x, y):
    """Return the product of x and y."""
    return x * y


def divide(x, y):
    """Return x divided by y."""
    return x / y


print("Select operation.")
print("1.Add")
print("2.Subtract")
print("3.Multiply")
print("4.Divide")

while True:
    choice = input("Enter choice(1/2/3/4): ")

    # Only accept a valid menu option.
    if choice in ('1', '2', '3', '4'):
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        # Bug fix: the original compared CHOICE, which was never defined.
        # Python names are case-sensitive, so this raised a NameError.
        # Once the indentation was consistent, this became the next
        # error Python reported.
        if choice == '1':
            print(num1, "+", num2, "=", add(num1, num2))
        elif choice == '2':
            print(num1, "-", num2, "=", subtract(num1, num2))
        elif choice == '3':
            print(num1, "*", num2, "=", multiply(num1, num2))
        elif choice == '4':
            print(num1, "/", num2, "=", divide(num1, num2))
        break
    else:
        # Bug fix: the original called Print(), which does not exist.
        # Built-in functions are lowercase, so this raised a NameError
        # whenever the user entered an invalid option.
        print("Invalid Input")
```
