```python
# View the data
import pandas as pd
data = {'State': ['Sao Paulo', 'Rio de Janeiro'],
        'Capital': ['Sao Paulo', 'Rio de Janeiro']}
bzl_df = pd.DataFrame(data)
print(data)
```

```python
new_info = pd.DataFrame({'State': ['Minas Gerais', 'Para'],
        'Capital': ['Belo Horizonte', 'Belem']})
bzl_df = pd.concat([bzl_df, new_info], ignore_index=True)

print(bzl_df.shape)
                
```
