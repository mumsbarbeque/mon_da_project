```python
import pandas as pd
# create a series of Students using student ID as index
class_list = pd.Series(['Thando', 'Divya', 'Simon', 'Peter'],
                       index=[30441216, 30441217, 30441218, 30441219])
class_list
```

```python
print(class_list.values)
print(class_list.index)
```

```python
# Retrieve a single value
class_list[30441218]
```

```python
# Retrieve multiple values with a list of index labels note the 2 brackets.
class_list[[30441218, 30441219]]
```

```python
# Recreate raw data as a dictionary
raw_data = {
    'Name': ['Thando', 'Divya', 'Simon', 'Peter'],
    'Country': ['South Africa', 'Singapore', 'United Kingdom', 'Australia'],
    'Qualification': ['PhD', 'Diploma', 'MBA', 'BSc'],
    'Age': [29, 20, 25, 23]}

# create a list for the row labels in the same order as the dictionary just created

row_labels = [30441216, 30441217, 30441218, 30441219]

# Create a datafram to with the new column data and explicit row labels

students_updated = pd.DataFrame(raw_data, columns = ['Name', 'Country', 'Qualification', 'Age',
             'Test 1', 'Tutorial 1', 'Test 2', 'Tutorial 2',
             'Test 3', 'Tutorial 3', 'Final mark', 'Pass'],
                               index=row_labels)
students_updated
```

```python
# assigning a single value to a column ie base line mark 80

students_updated['Test 1'] =80
students_updated
```

```python
# Assign values in order by using a list

students_updated['Test 1'] = [82, 65, 75, 90]
students_updated
```

```python
students_updated['Tutorial 2'] = [80, 90, 75, 68]

# Map original mark to new mark

tut_2_remark = {80: 84, 90: 88, 75: 79, 68: 74}
students_updated['tut_2_remark'] = students_updated['Tutorial 2'].map(tut_2_remark)
students_updated
```
